## Dataset A — Daily Order Values (₹)
1. **`mean()`** — Find the overall average order value for the month.
2. **`fmean()`** — Recompute the same average but as a precise float (compare speed/output type vs `mean()`).
3. **`geometric_mean()`** — Marketing claims order values "grew steadily." Compute the geometric mean and compare it to the arithmetic mean — which is lower, and why does that matter when averaging growth-like data?
4. **`median()`** — Find the median AOV. Is it higher or lower than the mean? What does that tell you about outliers (e.g., the ₹1600 day)?
5. **`median_low()` / `median_high()`** — Since there are 30 values (even count), find both the low and high median. Confirm they're different values.
6. **`mode()`** — Round every value down to the nearest ₹50 (e.g., 980 → 950) and store as `rounded_orders`. Find the mode of the rounded list.
7. **`multimode()`** — Using the same `rounded_orders` list, check if there's more than one mode.
8. **`variance()` / `pvariance()`** — Treat `order_values` as a *sample* of a larger population of all NovaMart orders. Calculate sample variance. Then, treat it as if it *is* the full population (all 30 days = entire month) and calculate population variance. Compare the two numbers.
9. **`stdev()` / `pstdev()`** — Calculate both standard deviations from the same list. Which is larger, and by how much?
10. **`quantiles()`** — Split `order_values` into 4 groups (quartiles) using `n=4`. What AOV separates the bottom 25% of days from the rest?

```python
Order_values = [
    980, 1050, 875, 1200, 999, 1450, 780, 1020, 1100, 950,
    1600, 890, 1010, 1075, 920, 1230, 1005, 860, 990, 1150,
    1300, 940, 1020, 1080, 970, 1500, 830, 1000, 1060, 910
]
```

```python
Average_Value = statistics.mean(Order_values)
print(Average_Value)
```

```python
fAverage_Value = statistics.fmean(Order_values)
print(fAverage_Value)
```

```python
Geomteric_mean = statistics.geometric_mean(Order_values)
print(Geomteric_mean)
```

```python
Median = statistics.median(Order_values)
print(Median)
```

```python
LMedian = statistics.median_low(Order_values)
print(LMedian)
HMedian = statistics.median_high(Order_values)
print(HMedian)
```

```python
Mode = statistics.mode(Order_values)
print(Mode)
```

```python
MMode = statistics.multimode(Order_values)
print(MMode)
```

```python
variance = statistics.variance(Order_values)
print(variance)
Pvariance = statistics.pvariance(Order_values)
print(Pvariance)
```

```python
SDeviation = statistics.stdev(Order_values)
print(SDeviation)
PDeviation = statistics.pstdev(Order_values)
print(PDeviation)
```

```python
Quantiles = statistics.quantiles(Order_values , n = 4)
print(Quantiles)
```
