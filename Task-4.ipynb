## Task 1

Your manager gives you the monthly salaries (in ₹ thousands) of 10 employees in the Sales team:

## Task 5
The number of support tickets closed per day by an employee over 9 days is:

`12, 15, 11, 20, 14, 13, 18, 16, 19`

Find the middle value of tickets closed.

Also, using a separate dataset of ticket counts for 8 days:

`10, 12, 14, 15, 18, 20, 22, 25`

Find both the lower-middle and higher-middle values.

## Task 6

The product IDs ordered by customers today are:

`101, 105, 101, 108, 101, 105, 110, 101`

Find the product that was ordered the most number of times.

Now, for a different dataset of defect codes found during quality checks:

`E1, E2, E1, E3, E2, E4, E1, E2`

Find all defect codes that occur with the highest frequency.

## Task 7

Your team collected the daily working hours of 6 employees (a sample from a larger department):

`8, 9, 7, 8.5, 9.5, 7.5`

Find how spread out these working hours are from the average, treating this as sample data.

Now treat the following dataset as the entire department's working hours (not a sample) and find the same measure:

`8, 9, 7, 8.5, 9.5, 7.5, 8, 9`

```python
Working_Hours_Sample = [8, 9, 7, 8.5, 9.5, 7.5]
Working_Hours = [8, 9, 7, 8.5, 9.5, 7.5, 8, 9]

Var = st.variance(Working_Hours_Sample)
print("Spread out these working hours are from the average is -",Var)

PVar_2 = st.pvariance(Working_Hours)
print("Spread out these working hours are from the average is -",PVar_2)
```

## Task 8

Two production lines report their daily output (units) over a week:

Line A (sample): `210, 215, 205, 220, 208, 212, 218`
Line B (entire population data): `300, 302, 298, 305, 299, 301, 303`

Find the standard deviation for Line A as sample data, and for Line B as population data. Based on the results, state which line has more consistent output.


```python
Line_A = [210, 215, 205, 220, 208, 212, 218]
Line_B = [300, 302, 298, 305, 299, 301, 303]

Std = st.stdev(Line_A)
print("The standard deviation for Line A as sample data is -",Std)

PStd = st.pstdev(Line_B)
print("The standard deviation for Line B as population data -",PStd)
```

## Task 11

The company wants to know if advertising spend affects sales. The data for 6 months is:

Advertising Spend (₹ thousands): `10, 15, 20, 25, 30, 35`
Sales (₹ thousands): `100, 130, 150, 170, 200, 220`

Find how these two variables change together, and also find how strongly they are related to each other.

```python
Advertising_Spend = [10, 15, 20, 25, 30, 35]
Sales = [100, 130, 150, 170, 200, 220]

Covariance = st.covariance(Advertising_Spend,Sales)
print("Covarienve is -",Covariance) 

Correlation = st.correlation(Advertising_Spend,Sales)
print("Correlation is -",Correlation)

#Conclusion: A correlation this close to 1 means spend and sales are very strongly related.
```

##  Task 12

Using the same advertising spend and sales data from Task 11, build a model that finds the best-fit line between advertising spend and sales. Using this model, predict the expected sales if the company spends ₹40,000 on advertising next month.

```python
Linear_Regression = st.linear_regression(Advertising_Spend,Sales)
print("A model that finds the best-fit line between advertising spend and sales -",Linear_Regression)

Prediction = Linear_Regression[1] + Linear_Regression[0] * 40000
print("The expected sales if the company spends ₹40,000 on advertising next month is -",Prediction)
```

## Task 14

A store sold the following quantities of 5 products today:

Quantities: `10, 5, 8, 12, 6`
Prices (₹ per unit): `50, 100, 75, 40, 60`

Find the total revenue generated by multiplying corresponding quantities and prices, then adding all the results together.


```python
Quantities = [10, 5, 8, 12, 6]
Price = [50, 100, 75, 40, 60]

Sum_Prod = st.sumprod(Quantities,Price)
print("The total revenue generated by multiplying corresponding quantities and prices, then adding all the results together is -",Sum_Prod)
```
