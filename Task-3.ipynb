```python
import statistics
```

## Sample Data

```python
# Daily sales (in ₹ thousands) for the Home Appliances section — 20 working days
Daily_sales = [42, 45, 39, 47, 44, 46, 41, 43, 45, 48,
               40, 44, 46, 42, 39, 240, 45, 43, 47, 44]

# Ad spend (in ₹ thousands) on the days that correspond to daily_sales above
Ad_spend = [5, 6, 4, 7, 6, 6, 5, 5, 6, 8,
            4, 6, 7, 5, 4, 6, 6, 5, 7, 6]

# Ratings (out of 5) given by 15 customers this week
Customer_ratings = [4, 5, 3, 4, 4, 5, 2, 4, 3, 5, 4, 4, 3, 5, 4]

# Delivery times in hours for 10 orders shipped to a nearby city
Delivery_hours = [3, 4, 3.5, 5, 4, 3, 12, 4.5, 3.5, 4]
```

### Part 1 — The "normal" day (`mean`, `fmean`, `median`, `mode`, `multimode`)

1. Find the average daily sales using `mean()` and `fmean()`.
2. Find the `median()` of `daily_sales`. Your manager wants to know: "what's a
   typical day look like?" — which measure, mean or median, answers that
   better here, and why?
3. Find the `mode()` (or `multimode()` if there's a tie) of `customer_ratings`.
   Write one sentence a manager could actually use: *"Most customers rate us
   a ___."*

```python
# 1
Mean = statistics.mean(Daily_sales)
print("The Mean of Daily Sales is -",Mean)

FMean = statistics.fmean(Daily_sales)
print("The FMean of Daily Sales is -",FMean)
```

```python
# 2
Median = statistics.median(Daily_sales)
print("The Median of Daily Sales is -",Median)

# Median (44) answers "typical day" better than mean (53.5) here,
# because the mean is being dragged upward by the 240 outlier. Median
# ignores the extreme value's magnitude and just looks at position.
```

```python
# 3
Mode = statistics.mode(Customer_ratings)
print("The Mode from Customer Rating is -",Mode)

Multimode = statistics.multimode(Customer_ratings)
print("The Multimode from Customer Rating is -",Multimode)

# Most Customer Rate as 4
```

### Part 2 — Spot the outlier (mean–median gap)

4. Compute the mean and median of `daily_sales`, then the **gap** between them
   (`mean - median`). Is the gap small or large compared to the typical
   day-to-day sales values?
5. Based on that gap, which single value in `daily_sales` do you suspect is a
   data entry error, and why? (Hint: someone probably typed an extra digit.)
6. Recompute the mean and median **after removing** that value. How much did
   the mean shift? Did the median shift at all — and what does that tell you
   about which measure is more resistant to outliers?

```python
# 4
Mean = statistics.mean(Daily_sales)
print("The Mean of Daily Sales is -",Mean)

Median = statistics.median(Daily_sales)
print("The Median of Daily Sales is -",Median)

Gap = Mean - Median 
print("The Gap is -",Gap)

# 9.5. Normal day-to-day sales only vary by a few units (39-48),
# so a gap of 9.5 is large relative to that spread -> outlier likely present.
```

```python
# 5
Outlier = max(Daily_sales)
print("The Suspect is -",Outlier)

# 240 stands out immediately -- every other value is in the 39-48 range.
# It looks like "24" got an extra 0 typed on the end.
```

```python
# 6
# After Removing Suspect
Clean_Daily_sales = [42, 45, 39, 47, 44, 46, 41, 43, 45, 48,
               40, 44, 46, 42, 39, 45, 43, 47, 44]

Mean = statistics.mean(Clean_Daily_sales)
print("The Mean of Daily Sales is -",Mean)

Median = statistics.median(Clean_Daily_sales)
print("The Median of Daily Sales is -",Median)

Gap = Mean - Median 
print("The Gap is -",Gap)

# Mean dropped from 53.5 to ~43.68 (a big shift). Median barely moved
# (44 -> 44). This shows median is resistant ("robust") to outliers while
# mean is sensitive to them -- a single bad data point can swing the mean
# a lot but rarely moves the median much.
```

### Part 3 — How consistent is the store? (`variance`, `stdev`, `pvariance`, `pstdev`)

7. Using the **cleaned** `daily_sales` (outlier removed), compute `variance()`
   and `stdev()`. Since these 19 days are a *sample* of the whole quarter, is
   `variance()`/`stdev()` the right pair, or should you use `pvariance()`/
   `pstdev()`? Justify your choice in one line.
8. Do the same for `delivery_hours` — find the outlier first (same trick as
   Part 2), clean it, then compute `stdev()`.

```python
# 7
Variance = statistics.variance(Clean_Daily_sales)
print("The Variance is -",Variance)

Stdev = statistics.stdev(Clean_Daily_sales)
print("The Standard Deviation is -",Stdev)

# These 19 days are a SAMPLE of the whole quarter (not every day that
# ever happened), so variance()/stdev() (sample versions) are correct.
# pvariance()/pstdev() would only be right if this were the entire
# population you cared about (e.g. all days that will ever exist).
```

```python
# 8
Mean_DH = statistics.mean(Delivery_hours)
print("The Mean of Delivery Hours is -",Mean_DH)

Median_DH = statistics.median(Delivery_hours)
print("The Median of Delivery Hours is -",Median_DH)

Gap = Mean_DH - Median_DH 
print("The Gap is -",Gap)

Outlier = max(Delivery_hours)
print("The Suspect is -",Outlier)

# Clean Data
Clean_Delivery_hours = [3, 4, 3.5, 5, 4, 3, 4.5, 3.5, 4]

Stdev = statistics.stdev(Clean_Delivery_hours)
print("The Standard Deviation is -",Stdev)
```

### Part 4 — Does ad spend actually help? (`covariance`, `correlation`, `linear regression`)

10. Is the correlation weak, moderate, or strong? Would you tell your manager
    "spend more on ads" based on this alone — why or why not? (Think about
    correlation ≠ causation from your notes.)

```python
Mean_AD = statistics.mean(Ad_spend)
print("The Mean of Ad Spend is -",Mean_AD)

Median_AD = statistics.median(Ad_spend)
print("The Median of Ad Spend is -",Median_AD)

Gap = Mean_AD - Median_AD 
print("The Gap is -",Gap)

Outlier = max(Ad_spend)
print("The Suspect is -",Outlier)

# Clean Data
Clean_Ad_spend = [5, 6, 4, 7, 6, 6, 5, 5, 6,
            4, 6, 7, 5, 4, 6, 6, 5, 7, 6]
```

```python
# 9
Covarianvce = statistics.covariance(Clean_Daily_sales,Clean_Ad_spend)
print("The Covariance is -",Covarianvce)

Correlation = statistics.correlation(Clean_Daily_sales,Clean_Ad_spend)
print("The Correlation is -",Correlation)
```

```python
# 10
# very strong positive relationship. It's
# tempting to say "spend more on ads," but correlation isn't causation:
# both could be driven by something else (e.g. a big sale weekend drives
# both foot traffic and ad spend up together)
```
