## Task - 1 
1. One of these warehouses has a data-entry error (an obvious outlier). Identify which one and which value, using the **gap between `mean()` and `median()`** as your evidence — don't just eyeball it, show the numeric gap for all three warehouses.
2. Recompute `north`'s mean *after* removing the suspected error. Report both the "before" and "after" mean, and the % change.
3. For all three warehouses, compute `mode()` and `multimode()`. One warehouse will return a tie (more than one mode) — identify it and explain in one line what a multi-modal result implies about that warehouse's pricing (e.g., two common price points vs. one).
4. Compute `stdev()` for all three warehouses (use the cleaned `north` list from Task 2). Rank the three warehouses from most to least "consistent" in order value.
5. `south` and `east` have different means and different spreads. Using `stdev()` alone, which one would you tell your manager is "more predictable" — and is that the same as saying it's "better performing"? Justify in 1–2 lines.

```python
North = [1200, 1180, 1250, 1200, 1195, 1220, 1200, 3400, 1210, 1190]
South = [860, 875, 900, 860, 880, 895, 860, 910, 875, 900, 880]
East  = [1500, 1520, 1495, 1510, 1505, 1500, 1518, 1500, 1495]
```

```python
Line = "North Region"
print(Line.center(40,"-"))

North_Mean = statistics.mean(North)
print("North Mean -",North_Mean)
North_Median = statistics.median(North)
print("North Median - ",North_Median)
Gap = North_Mean - North_Median
print("Gap - " ,Gap)

Line = "South Region"
print(Line.center(40,"-"))

South_Mean = statistics.mean(South)
print("South Mean -",South_Mean)
South_Median = statistics.median(South)
print("South Median -",South_Median)
Gap = South_Mean - South_Median
print("Gap -",Gap)

Line = "East Region"
print(Line.center(40,"-"))

East_Mean = statistics.mean(East)
print("East Mean -",East_Mean)
East_Median = statistics.median(East)
print("East Median -",East_Mean)
Gap = East_Mean - East_Mean
print("Gap -",Gap)
```

```python
Line = "North Mean Before Removing Error"
print(Line.center(40,"-"))

North = [1200, 1180, 1250, 1200, 1195, 1220, 1200, 3400, 1210, 1190]
North_Mean_Before = statistics.mean(North)
print("Mean -",North_Mean_Before)

Line = "North Mean After Removing Error"
print(Line.center(40,"-"))

North_Clean = [1200, 1180, 1250, 1200, 1195, 1220, 1200, 1210, 1190]
North_Mean_After = statistics.mean(North_Clean)
print("Mean -",North_Mean_After)

Line = "Percentage Change"
print(Line.center(40,"-"))

Percentage_change = ((North_Mean_Before - North_Mean_After) / North_Mean_Before) * 100
print("% Change -", Percentage_change)
```

```python
Line = "North Region"
print(Line.center(40,"-"))

North_Mode = statistics.mode(North)
print("North Mode -",North_Mode)
North_Multimode = statistics.multimode(North)
print("North Multimode - ",North_Multimode)

Line = "South Region"
print(Line.center(40,"-"))

South_Mode = statistics.mode(South)
print("South Mode -",South_Mode)
South_Multimode = statistics.multimode(South)
print("South Multimode - ",South_Multimode)

Line = "East Region"
print(Line.center(40,"-"))

East_Mode = statistics.mode(East)
print("East Mode -",East_Mode)
East_Multimode = statistics.multimode(East)
print("East Multimode - ",East_Multimode)

print("No Warehouse is Tie")
```

```python
Line = "North Region"
print(Line.center(40,"-"))

North_std = statistics.stdev(North_Clean)
print("Standard Deviation -",North_std)

Line = "South Region"
print(Line.center(40,"-"))

South_std = statistics.stdev(South)
print("Standard Deviation -",South_std)

Line = "East Region"
print(Line.center(40,"-"))

East_std = statistics.stdev(East)
print("Standard Deviation -",East_std)

Line = "Ranking"
print(Line.center(40,"-"))

print("1.East Region")
print("2.South Region")
print("3.North Region")
```

```python
Line = "South Region"
print(Line.center(40,"-"))

South_Mean = statistics.mean(South)
print("South Mean -",South_Mean)
South_std = statistics.stdev(South)
print("Standard Deviation -",South_std)

Line = "East Region"
print(Line.center(40,"-"))

East_Mean = statistics.mean(East)
print("East Mean -",East_Mean)
East_std = statistics.stdev(East)
print("Standard Deviation -",East_std)

Line = "Which is More Predictable ?"
print(Line.center(40,"-"))
print("1 - East is more predictable because its standard deviation is lower")
print("2 - Standard deviation only tells us about consistency/spread, not whether the warehouse is performing better")
```

```python
weeks         = list(range(1, 9))
Email_spend   = [2, 2, 3, 3, 4, 4, 5, 5]       # ₹ hundreds
Email_orders  = [30, 28, 38, 41, 44, 46, 52, 55]

Social_spend  = [5, 6, 4, 7, 3, 8, 2, 9]
Social_orders = [60, 58, 62, 55, 68, 50, 70, 45]

Influencer_spend  = [1, 1, 1, 2, 2, 2, 3, 3]
Influencer_orders = [15, 18, 14, 25, 27, 22, 5, 40]
```

```python
Line = "Email Coverage"
print(Line.center(40,"-"))

Correlation_Email = statistics.correlation(Email_spend,Email_orders)
print("Correlation -",Correlation_Email)

Line = "Social Coverage"
print(Line.center(40,"-"))

Correlation_Social = statistics.correlation(Social_spend,Social_orders)
print("Correlation -",Correlation_Social)

Line = "Influencer Coverage"
print(Line.center(40,"-"))

Correlation_Influencer = statistics.correlation(Influencer_spend,Influencer_orders)
print("Correlation -",Correlation_Influencer)

Line = "Ranking"
print(Line.center(40,"-"))

print("1.Email")
print("2.Influencer")
print("2.Social")
```

```python
Line = "Checking Correlation and Covarince of Social Coverage"
print(Line.center(70,"-"))

Correlation_Social = statistics.correlation(Social_spend,Social_orders)
print("Correlation -",Correlation_Social)

Covariance_Social = statistics.covariance(Social_spend,Social_orders)
print("Covariance -",Covariance_Social)

print("Covariance is negative: This tells us that, in general: When social spend increases, social orders tend to decrease")
```

```python
Line = "Linear regression on strongest positive correlation"
print(Line.center(70,"-"))

Linear_Regression = statistics.linear_regression(Email_spend,Email_orders)
print(Linear_Regression)
```

```python
Line = "Checking Variance of Influencer"
print(Line.center(70,"-"))

Variance_Inf = statistics.variance(Influencer_orders)
print("Variance -",Variance_Inf)

Mean_Inf = statistics.mean(Influencer_orders)
print("Mean -",Mean_Inf)

print("Variance is actually High , as we compare to the spend")
```

```python
Agent_priya = [9, 8, 9, 10, 9, 8, 9]
Agent_ravi  = [7, 8, 6, 9, 7]
Agent_meena = [9, 2, 10, 9, 10, 8, 9, 10, 1, 9, 10]
```

```python
Line = "Agent Priya"
print(Line.center(40,"-"))

Mean_Priya = statistics.mean(Agent_priya)
print("Mean -",Mean_Priya)
Median_Priya = statistics.median(Agent_priya)
print("Median -",Median_Priya)

Line = "Agent Ravi"
print(Line.center(40,"-"))

Mean_Ravi = statistics.mean(Agent_ravi)
print("Mean -",Mean_Ravi)
Median_Ravi = statistics.median(Agent_ravi)
print("Median -",Median_Ravi)

Line = "Agent Meena"
print(Line.center(40,"-"))

Mean_Meena = statistics.mean(Agent_meena)
print("Mean -",Mean_Meena)
Median_Meena = statistics.median(Agent_meena)
print("Median -",Median_Meena)

Line = "Standard Deviation"
print(Line.center(40,"-"))

Stdv_Priya = statistics.stdev(Agent_priya)
print("Standard Deviation (Priya) -",Stdv_Priya)

Stdv_Ravi = statistics.stdev(Agent_ravi)
print("Standard Deviation (Ravi) -",Stdv_Ravi)

Stdv_Meena = statistics.stdev(Agent_meena)
print("Standard Deviation (Meena) -",Stdv_Meena)

print("Meena's mean (7.91) is lower than her median (9) because the two unusually low scores, 1 and 2, pull the mean downward")
```
